```python
# ============================================================
# SEC FILING NLP ALPHA ENGINE
# University-Level Quantitative Finance Research Project
# ============================================================
#
# Research question:
# Can information extracted from SEC filings using NLP
# help explain short-term stock returns?
#
# Methods:
# - SEC EDGAR filings
# - FinBERT sentiment analysis
# - Risk-language intensity
# - Filing novelty
# - Sentiment changes
# - 5-day forward stock returns
# - Portfolio sorting
# - Cross-sectional regression
# - Simple long-short strategy
#
# This is a research prototype, NOT a live trading system.
# ============================================================

# ============================================================
# CLEAN PYTORCH / TRANSFORMERS SETUP
# ============================================================

!pip -q install -U transformers

import torch

print("PyTorch version:", torch.__version__)
print(
    "Device:",
    "CUDA" if torch.cuda.is_available() else "CPU"
)

from transformers import AutoTokenizer, AutoModelForSequenceClassification

MODEL_NAME = "ProsusAI/finbert"

tokenizer = AutoTokenizer.from_pretrained(
    MODEL_NAME
)

model = AutoModelForSequenceClassification.from_pretrained(
    MODEL_NAME
)

device = torch.device(
    "cuda" if torch.cuda.is_available() else "cpu"
)

model.to(device)
model.eval()

print("FinBERT loaded successfully.")
print("Running on:", device)

# ============================================================
# CELL 1 — INSTALL PACKAGES
# ============================================================

!pip -q install transformers yfinance beautifulsoup4 lxml statsmodels seaborn


# ============================================================
# CELL 2 — IMPORTS
# ============================================================

import os
import re
import time
import math
import warnings
import requests
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import statsmodels.api as sm
import yfinance as yf

from bs4 import BeautifulSoup
from collections import Counter
from datetime import datetime, timedelta

import torch
from transformers import AutoTokenizer, AutoModelForSequenceClassification


warnings.filterwarnings("ignore")

pd.set_option("display.max_columns", 100)
pd.set_option("display.width", 140)

print("Libraries loaded successfully.")
print("PyTorch device:", "CUDA" if torch.cuda.is_available() else "CPU")


# ============================================================
# CELL 3 — PROJECT CONFIGURATION
# ============================================================

PROJECT_NAME = "SEC Filing NLP Alpha Engine"

START_DATE = "2020-01-01"
END_DATE = "2026-09-01"

# Keep this reasonably small for a university project.
# You can expand it later.
TICKERS = [
    "AAPL",
    "MSFT",
    "AMZN",
    "GOOGL",
    "META",
    "NVDA",
    "JPM",
    "BAC",
    "GS",
    "MS",
    "XOM",
    "CVX",
    "WMT",
    "COST",
    "KO",
    "PEP",
    "JNJ",
    "PFE",
    "UNH",
    "HD"
]

MAX_FILINGS = 250

FORWARD_DAYS = 5

print("Project:", PROJECT_NAME)
print("Number of stocks:", len(TICKERS))
print("Maximum filings:", MAX_FILINGS)


# ============================================================
# CELL 4 — DIRECTORIES
# ============================================================

BASE_DIR = "/content/sec_nlp_project"

DATA_DIR = os.path.join(BASE_DIR, "data")
OUTPUT_DIR = os.path.join(BASE_DIR, "outputs")

os.makedirs(DATA_DIR, exist_ok=True)
os.makedirs(OUTPUT_DIR, exist_ok=True)

print("Project directory:", BASE_DIR)


# ============================================================
# CELL 5 — SEC CONFIGURATION
# ============================================================
#
# SEC requires a descriptive User-Agent.
# Replace the email below with your own email if desired.
# ============================================================

SEC_HEADERS = {
    "User-Agent": "University Research Project student@example.com"
}

SEC_TICKERS_URL = "https://www.sec.gov/files/company_tickers.json"

session = requests.Session()
session.headers.update(SEC_HEADERS)

print("SEC session configured.")


# ============================================================
# CELL 6 — DOWNLOAD SEC TICKER DATABASE
# ============================================================

response = session.get(SEC_TICKERS_URL)

if response.status_code != 200:
    raise RuntimeError(
        f"SEC ticker database request failed: {response.status_code}"
    )

ticker_data = response.json()

ticker_lookup = {}

for item in ticker_data.values():
    ticker = item["ticker"].upper()
    cik = str(item["cik_str"]).zfill(10)
    ticker_lookup[ticker] = cik

print("SEC ticker database loaded.")
print("Number of companies:", len(ticker_lookup))


# ============================================================
# CELL 7 — GET FILING METADATA
# ============================================================

def get_company_filings(ticker):

    ticker = ticker.upper()

    if ticker not in ticker_lookup:
        return pd.DataFrame()

    cik = ticker_lookup[ticker]

    url = f"https://data.sec.gov/submissions/CIK{cik}.json"

    response = session.get(url)

    if response.status_code != 200:
        print(f"{ticker}: SEC request failed ({response.status_code})")
        return pd.DataFrame()

    data = response.json()

    recent = pd.DataFrame(data["filings"]["recent"])

    if recent.empty:
        return pd.DataFrame()

    recent = recent[
        recent["form"].isin(["10-K", "10-Q"])
    ].copy()

    recent["ticker"] = ticker

    return recent


all_filings = []

for ticker in TICKERS:

    print("Downloading metadata:", ticker)

    try:
        df = get_company_filings(ticker)

        if not df.empty:
            all_filings.append(df)

    except Exception as e:
        print("Error:", ticker, e)

    time.sleep(0.15)


filings = pd.concat(
    all_filings,
    ignore_index=True
)

print("\nTotal filings retrieved:", len(filings))


# ============================================================
# CELL 8 — FILTER DATES
# ============================================================

filings["filingDate"] = pd.to_datetime(
    filings["filingDate"],
    errors="coerce"
)

filings = filings[
    (filings["filingDate"] >= START_DATE) &
    (filings["filingDate"] <= END_DATE)
].copy()

filings = filings.sort_values(
    ["ticker", "filingDate"]
).reset_index(drop=True)

print("Filings after date filter:", len(filings))

display(
    filings[
        [
            "ticker",
            "form",
            "filingDate",
            "accessionNumber",
            "primaryDocument"
        ]
    ].head(10)
)


# ============================================================
# CELL 9 — BUILD SEC DOCUMENT URLS
# ============================================================

def build_sec_url(row):

    cik = ticker_lookup[row["ticker"]]

    accession = row["accessionNumber"].replace("-", "")

    document = row["primaryDocument"]

    return (
        f"https://www.sec.gov/Archives/edgar/data/"
        f"{int(cik)}/{accession}/{document}"
    )


filings["url"] = filings.apply(
    build_sec_url,
    axis=1
)

print(filings[["ticker", "form", "filingDate", "url"]].head())


# ============================================================
# CELL 10 — LIMIT DATASET
# ============================================================

# Keep the most recent MAX_FILINGS observations.
# This keeps the notebook manageable.
filings = (
    filings
    .sort_values("filingDate", ascending=False)
    .head(MAX_FILINGS)
    .sort_values(["ticker", "filingDate"])
    .reset_index(drop=True)
)

print("Final filing sample:", len(filings))

print(
    filings["ticker"].value_counts().sort_index()
)


# ============================================================
# CELL 11 — DOWNLOAD AND CLEAN SEC FILINGS
# ============================================================

def download_filing(url):

    try:

        response = session.get(url, timeout=30)

        if response.status_code != 200:
            return ""

        return response.text

    except Exception:
        return ""


def clean_html(html):

    if not isinstance(html, str) or len(html) == 0:
        return ""

    soup = BeautifulSoup(html, "lxml")

    # Remove scripts/styles
    for tag in soup(["script", "style", "noscript"]):
        tag.decompose()

    text = soup.get_text(" ")

    # Normalize whitespace
    text = re.sub(r"\s+", " ", text)

    # Remove excessive spaces
    text = text.strip()

    return text


filing_texts = []

for i, row in filings.iterrows():

    print(
        f"Downloading {i+1}/{len(filings)} "
        f"{row['ticker']} {row['form']}"
    )

    html = download_filing(row["url"])

    text = clean_html(html)

    filing_texts.append(text)

    time.sleep(0.15)


filings["clean_text"] = filing_texts

filings["text_length"] = (
    filings["clean_text"]
    .str.len()
)

print("\nSuccessfully downloaded:",
      (filings["text_length"] > 0).sum())

print(
    filings["text_length"].describe()
)


# ============================================================
# CELL 12 — REMOVE VERY SHORT / INVALID FILINGS
# ============================================================

filings = filings[
    filings["text_length"] >= 5000
].copy()

filings = filings.reset_index(drop=True)

print("Valid filings:", len(filings))


# ============================================================
# CELL 13 — BASIC TEXT PROCESSING
# ============================================================

def normalize_text(text):

    if not isinstance(text, str):
        return ""

    text = text.lower()

    text = re.sub(
        r"\s+",
        " ",
        text
    )

    return text.strip()


filings["normalized_text"] = (
    filings["clean_text"]
    .apply(normalize_text)
)


# ============================================================
# CELL 14 — LINGUISTIC FEATURES
# ============================================================

def count_words(text):

    if not isinstance(text, str):
        return 0

    return len(
        re.findall(
            r"\b\w+\b",
            text
        )
    )


def count_sentences(text):

    if not isinstance(text, str):
        return 0

    sentences = re.split(
        r"[.!?]+",
        text
    )

    return len(
        [
            s for s in sentences
            if len(s.strip()) > 10
        ]
    )


filings["word_count"] = (
    filings["normalized_text"]
    .apply(count_words)
)

filings["sentence_count"] = (
    filings["normalized_text"]
    .apply(count_sentences)
)

filings["avg_words_per_sentence"] = (
    filings["word_count"] /
    filings["sentence_count"].replace(0, np.nan)
)

filings["log_word_count"] = np.log1p(
    filings["word_count"]
)

print(
    filings[
        [
            "ticker",
            "word_count",
            "sentence_count",
            "avg_words_per_sentence"
        ]
    ].head()
)


# ============================================================
# CELL 15 — RISK LANGUAGE FEATURE
# ============================================================

RISK_TERMS = [
    "risk",
    "risks",
    "uncertain",
    "uncertainty",
    "uncertainties",
    "volatile",
    "volatility",
    "adverse",
    "decline",
    "declines",
    "declining",
    "loss",
    "losses",
    "litigation",
    "lawsuit",
    "lawsuits",
    "regulatory",
    "regulation",
    "regulations",
    "compliance",
    "default",
    "defaults",
    "liquidity",
    "cyberattack",
    "cybersecurity",
    "disruption",
    "disruptions",
    "inflation",
    "recession",
    "downturn",
    "geopolitical",
    "sanctions",
    "impairment",
    "impairments",
    "material weakness",
    "going concern"
]


def calculate_risk_language(text):

    if not isinstance(text, str):
        return np.nan

    text = text.lower()

    words = re.findall(
        r"\b\w+\b",
        text
    )

    if len(words) == 0:
        return np.nan

    risk_count = 0

    for term in RISK_TERMS:

        pattern = (
            r"\b" +
            re.escape(term) +
            r"\b"
        )

        risk_count += len(
            re.findall(
                pattern,
                text
            )
        )

    # occurrences per 1,000 words
    return (
        risk_count /
        len(words)
    ) * 1000


filings["risk_language"] = (
    filings["normalized_text"]
    .apply(calculate_risk_language)
)

print(
    filings["risk_language"].describe()
)


# ============================================================
# CELL 16 — LOAD FINBERT
# ============================================================

MODEL_NAME = "ProsusAI/finbert"

device = torch.device(
    "cuda"
    if torch.cuda.is_available()
    else "cpu"
)

print("Loading FinBERT...")
print("Device:", device)

tokenizer = AutoTokenizer.from_pretrained(
    MODEL_NAME
)

model = AutoModelForSequenceClassification.from_pretrained(
    MODEL_NAME
)

model.to(device)

model.eval()

print("FinBERT loaded successfully.")


# ============================================================
# CELL 17 — FINBERT SENTIMENT FUNCTION
# ============================================================

def finbert_score_text(
    text,
    max_sentences=80
):

    if not isinstance(text, str):
        return np.nan

    # Split filing into sentences
    sentences = re.split(
        r"(?<=[.!?])\s+",
        text
    )

    sentences = [
        s.strip()
        for s in sentences
        if len(s.strip()) > 30
    ]

    if len(sentences) == 0:
        return np.nan

    # Sample long filings to keep runtime manageable
    if len(sentences) > max_sentences:

        indices = np.linspace(
            0,
            len(sentences) - 1,
            max_sentences
        ).astype(int)

        sentences = [
            sentences[i]
            for i in indices
        ]

    scores = []

    batch_size = 16

    for start in range(
        0,
        len(sentences),
        batch_size
    ):

        batch = sentences[
            start:start + batch_size
        ]

        encoded = tokenizer(
            batch,
            padding=True,
            truncation=True,
            max_length=256,
            return_tensors="pt"
        )

        encoded = {
            k: v.to(device)
            for k, v in encoded.items()
        }

        with torch.no_grad():

            output = model(
                **encoded
            )

            probabilities = torch.softmax(
                output.logits,
                dim=1
            )

        # FinBERT labels:
        # 0 = positive
        # 1 = negative
        # 2 = neutral

        positive = probabilities[:, 0]
        negative = probabilities[:, 1]

        batch_scores = (
            positive -
            negative
        )

        scores.extend(
            batch_scores.detach()
            .cpu()
            .numpy()
        )

    return float(
        np.mean(scores)
    )


# ============================================================
# CELL 18 — RUN FINBERT
# ============================================================

sentiment_scores = []

for i, text in enumerate(
    filings["clean_text"]
):

    print(
        f"FinBERT: {i+1}/{len(filings)}"
    )

    try:

        score = finbert_score_text(
            text
        )

    except Exception as e:

        print(
            "Error:",
            e
        )

        score = np.nan

    sentiment_scores.append(
        score
    )


filings["sentiment_score"] = (
    sentiment_scores
)

print(
    filings[
        [
            "ticker",
            "filingDate",
            "sentiment_score"
        ]
    ].head()
)


# ============================================================
# CELL 19 — SENTIMENT CHANGE
# ============================================================

filings = filings.sort_values(
    ["ticker", "filingDate"]
).copy()

filings["sentiment_change"] = (
    filings
    .groupby("ticker")["sentiment_score"]
    .diff()
)

filings["risk_change"] = (
    filings
    .groupby("ticker")["risk_language"]
    .diff()
)


# ============================================================
# CELL 20 — FILING NOVELTY
# ============================================================
#
# Simple novelty measure:
# Compare current filing vocabulary with previous filing.
#
# novelty = 1 - Jaccard similarity
#
# Higher value = more different from previous filing.
# ============================================================

def calculate_novelty(current, previous):

    if not isinstance(current, str):
        return np.nan

    if not isinstance(previous, str):
        return np.nan

    current_words = set(
        re.findall(
            r"\b[a-z]{4,}\b",
            current.lower()
        )
    )

    previous_words = set(
        re.findall(
            r"\b[a-z]{4,}\b",
            previous.lower()
        )
    )

    if len(current_words) == 0:
        return np.nan

    union = (
        current_words |
        previous_words
    )

    intersection = (
        current_words &
        previous_words
    )

    if len(union) == 0:
        return np.nan

    similarity = (
        len(intersection) /
        len(union)
    )

    return 1 - similarity


filings["previous_text"] = (
    filings
    .groupby("ticker")["normalized_text"]
    .shift(1)
)

filings["novelty"] = filings.apply(
    lambda row:
        calculate_novelty(
            row["normalized_text"],
            row["previous_text"]
        ),
    axis=1
)

print(
    filings[
        [
            "ticker",
            "filingDate",
            "novelty"
        ]
    ].head(15)
)


# ============================================================
# CELL 21 — PREPARE RESEARCH DATASET
# ============================================================

research_df = filings[
    [
        "ticker",
        "form",
        "filingDate",
        "accessionNumber",
        "sentiment_score",
        "sentiment_change",
        "risk_language",
        "risk_change",
        "novelty",
        "word_count",
        "log_word_count"
    ]
].copy()

research_df = research_df.rename(
    columns={
        "filingDate": "date"
    }
)

research_df = research_df.sort_values(
    ["ticker", "date"]
).reset_index(drop=True)

print(
    "Research dataset:",
    research_df.shape
)

display(
    research_df.head()
)


# ============================================================
# CELL 22 — DOWNLOAD STOCK PRICES
# ============================================================

price_data = {}

for ticker in TICKERS:

    print(
        "Downloading prices:",
        ticker
    )

    try:

        data = yf.download(
            ticker,
            start="2019-12-01",
            end="2026-10-01",
            auto_adjust=True,
            progress=False
        )

        if data.empty:
            print(
                "No data:",
                ticker
            )
            continue

        # Handle possible MultiIndex
        if isinstance(
            data.columns,
            pd.MultiIndex
        ):

            close = data[
                "Close"
            ]

            if isinstance(
                close,
                pd.DataFrame
            ):
                close = close.iloc[:, 0]

        else:

            close = data["Close"]

        close.index = pd.to_datetime(
            close.index
        )

        price_data[ticker] = (
            close.dropna()
        )

    except Exception as e:

        print(
            "Price error:",
            ticker,
            e
        )


print(
    "Stocks with price data:",
    len(price_data)
)


# ============================================================
# CELL 23 — FORWARD RETURN FUNCTION
# ============================================================

def calculate_forward_return(
    ticker,
    event_date,
    horizon=5
):

    if ticker not in price_data:
        return np.nan

    prices = price_data[ticker]

    event_date = pd.Timestamp(
        event_date
    )

    # Use the first trading day ON OR AFTER
    # the filing date.
    future_dates = prices.index[
        prices.index >= event_date
    ]

    if len(future_dates) <= horizon:
        return np.nan

    entry_date = future_dates[0]

    exit_date = future_dates[horizon]

    entry_price = prices.loc[
        entry_date
    ]

    exit_price = prices.loc[
        exit_date
    ]

    if (
        pd.isna(entry_price) or
        pd.isna(exit_price) or
        entry_price <= 0
    ):
        return np.nan

    return (
        exit_price /
        entry_price
    ) - 1


# ============================================================
# CELL 24 — CALCULATE FORWARD RETURNS
# ============================================================

research_df["ret_5d"] = research_df.apply(
    lambda row:
        calculate_forward_return(
            row["ticker"],
            row["date"],
            FORWARD_DAYS
        ),
    axis=1
)

print(
    research_df["ret_5d"].describe()
)


# ============================================================
# CELL 25 — CLEAN FINAL RESEARCH SAMPLE
# ============================================================

analysis_columns = [
    "ticker",
    "form",
    "date",
    "sentiment_score",
    "sentiment_change",
    "risk_language",
    "risk_change",
    "novelty",
    "log_word_count",
    "ret_5d"
]

research_df = research_df[
    analysis_columns
].copy()

research_df = research_df.replace(
    [np.inf, -np.inf],
    np.nan
)

research_df = research_df.dropna(
    subset=[
        "ret_5d",
        "sentiment_score",
        "risk_language",
        "novelty"
    ]
).copy()

research_df = research_df.sort_values(
    "date"
).reset_index(drop=True)

print(
    "Final research sample:",
    research_df.shape
)

display(
    research_df.head()
)


# ============================================================
# CELL 26 — DESCRIPTIVE STATISTICS
# ============================================================

print(
    research_df[
        [
            "sentiment_score",
            "sentiment_change",
            "risk_language",
            "risk_change",
            "novelty",
            "ret_5d"
        ]
    ].describe()
)


# ============================================================
# CELL 27 — SENTIMENT DISTRIBUTION
# ============================================================

plt.figure(
    figsize=(9, 5)
)

plt.hist(
    research_df["sentiment_score"],
    bins=30
)

plt.axvline(
    0,
    linestyle="--"
)

plt.title(
    "Distribution of FinBERT Filing Sentiment"
)

plt.xlabel(
    "Sentiment Score"
)

plt.ylabel(
    "Number of Filings"
)

plt.tight_layout()

plt.show()


# ============================================================
# CELL 28 — SENTIMENT VS FORWARD RETURNS
# ============================================================

plt.figure(
    figsize=(9, 6)
)

plt.scatter(
    research_df["sentiment_score"],
    research_df["ret_5d"],
    alpha=0.6
)

plt.axhline(
    0,
    linestyle="--"
)

plt.axvline(
    0,
    linestyle="--"
)

plt.title(
    "Filing Sentiment vs 5-Day Forward Return"
)

plt.xlabel(
    "FinBERT Sentiment"
)

plt.ylabel(
    "5-Day Forward Return"
)

plt.tight_layout()

plt.show()


# ============================================================
# CELL 29 — SENTIMENT QUINTILES
# ============================================================

research_df["sentiment_quantile"] = (
    research_df.groupby("date")[
        "sentiment_score"
    ]
    .transform(
        lambda x:
        pd.qcut(
            x,
            5,
            labels=False,
            duplicates="drop"
        )
    )
)

quantile_returns = (
    research_df
    .groupby("sentiment_quantile")[
        "ret_5d"
    ]
    .mean()
)

print(
    "Average 5-day return by sentiment quintile:"
)

print(
    quantile_returns
)

if (
    0 in quantile_returns.index and
    4 in quantile_returns.index
):

    spread = (
        quantile_returns.loc[4]
        -
        quantile_returns.loc[0]
    )

    print(
        f"\nHigh-minus-low sentiment spread: "
        f"{spread:.4%}"
    )


# ============================================================
# CELL 30 — QUANTILE PLOT
# ============================================================

plt.figure(
    figsize=(9, 5)
)

quantile_returns.plot(
    kind="bar"
)

plt.axhline(
    0,
    linestyle="--"
)

plt.title(
    "Average 5-Day Return by Sentiment Quintile"
)

plt.xlabel(
    "Sentiment Quintile"
)

plt.ylabel(
    "Average 5-Day Return"
)

plt.tight_layout()

plt.show()


# ============================================================
# CELL 31 — CROSS-SECTIONAL REGRESSION
# ============================================================

regression_df = research_df[
    [
        "ret_5d",
        "sentiment_score",
        "sentiment_change",
        "novelty",
        "risk_language",
        "log_word_count"
    ]
].copy()

regression_df = regression_df.replace(
    [np.inf, -np.inf],
    np.nan
)

regression_df = regression_df.dropna()

X = regression_df[
    [
        "sentiment_score",
        "sentiment_change",
        "novelty",
        "risk_language",
        "log_word_count"
    ]
]

y = regression_df[
    "ret_5d"
]

X = sm.add_constant(X)

model = sm.OLS(
    y,
    X
).fit(
    cov_type="HC3"
)

print(
    model.summary()
)


# ============================================================
# CELL 32 — CREATE COMPOSITE NLP SIGNAL
# ============================================================
#
# Standardize signals so they are comparable.
#
# Positive:
#   sentiment
#   sentiment change
#   novelty
#
# Negative:
#   risk language
#
# The weights are intentionally simple rather than optimized.
# ============================================================

signal_df = research_df.copy()


def zscore(series):

    std = series.std()

    if std == 0 or pd.isna(std):
        return pd.Series(
            0,
            index=series.index
        )

    return (
        series - series.mean()
    ) / std


signal_df["z_sentiment"] = (
    zscore(
        signal_df["sentiment_score"]
    )
)

signal_df["z_sentiment_change"] = (
    zscore(
        signal_df["sentiment_change"]
    )
)

signal_df["z_novelty"] = (
    zscore(
        signal_df["novelty"]
    )
)

signal_df["z_risk"] = (
    zscore(
        signal_df["risk_language"]
    )
)

signal_df["alpha_signal"] = (
    0.50 *
    signal_df["z_sentiment"]
    +
    0.25 *
    signal_df["z_sentiment_change"]
    +
    0.25 *
    signal_df["z_novelty"]
    -
    0.25 *
    signal_df["z_risk"]
)

print(
    signal_df[
        [
            "ticker",
            "date",
            "alpha_signal"
        ]
    ].head()
)


# ============================================================
# CELL 33 — FORM PORTFOLIOS BY FILING DATE
# ============================================================
#
# Important:
# Each filing date forms a cross-sectional portfolio.
#
# Long:
#   Highest-alpha observations
#
# Short:
#   Lowest-alpha observations
#
# Portfolio return:
#   Long return - Short return
#
# We require at least 4 observations on a date.
# This prevents empty portfolios.
# ============================================================

portfolio_records = []

for date, group in signal_df.groupby("date"):

    group = group.dropna(
        subset=[
            "alpha_signal",
            "ret_5d"
        ]
    ).copy()

    # Need enough stocks to form long/short portfolios
    if len(group) < 4:
        continue

    group = group.sort_values(
        "alpha_signal"
    )

    # Bottom 25%
    n_long_short = max(
        1,
        int(len(group) * 0.25)
    )

    short_group = group.head(
        n_long_short
    )

    long_group = group.tail(
        n_long_short
    )

    long_return = (
        long_group["ret_5d"]
        .mean()
    )

    short_return = (
        short_group["ret_5d"]
        .mean()
    )

    long_short_return = (
        long_return -
        short_return
    )

    portfolio_records.append(
        {
            "date": date,
            "long_return": long_return,
            "short_return": short_return,
            "long_short_return": long_short_return,
            "n_stocks": len(group),
            "n_long": len(long_group),
            "n_short": len(short_group)
        }
    )


# ============================================================
# CELL 34 — CREATE PORTFOLIO DATAFRAME
# ============================================================

portfolio = pd.DataFrame(
    portfolio_records
)

if portfolio.empty:

    raise ValueError(
        "The portfolio is empty. "
        "There are not enough cross-sectional observations "
        "on the filing dates to form portfolios."
    )

portfolio["date"] = pd.to_datetime(
    portfolio["date"]
)

portfolio = portfolio.sort_values(
    "date"
).reset_index(drop=True)

print(
    "Portfolio observations:",
    len(portfolio)
)

display(
    portfolio.head(10)
)


# ============================================================
# CELL 35 — CUMULATIVE LONG-SHORT PERFORMANCE
# ============================================================

portfolio["long_short_growth"] = (
    1 +
    portfolio["long_short_return"]
).cumprod()

portfolio["long_growth"] = (
    1 +
    portfolio["long_return"]
).cumprod()

# For an actual short P&L approximation:
# short position gains when the underlying short basket falls.
portfolio["short_pnl"] = (
    -portfolio["short_return"]
)

portfolio["short_growth"] = (
    1 +
    portfolio["short_pnl"]
).cumprod()


# ============================================================
# CELL 36 — PLOT PORTFOLIO PERFORMANCE
# ============================================================

plt.figure(
    figsize=(11, 6)
)

plt.plot(
    portfolio["date"],
    portfolio["long_short_growth"],
    label="Long-Short"
)

plt.plot(
    portfolio["date"],
    portfolio["long_growth"],
    label="Long Portfolio"
)

plt.title(
    "SEC NLP Signal Portfolio Performance"
)

plt.xlabel(
    "Date"
)

plt.ylabel(
    "Growth of £1"
)

plt.legend()

plt.grid(
    alpha=0.3
)

plt.tight_layout()

plt.show()


# ============================================================
# CELL 37 — PORTFOLIO PERFORMANCE STATISTICS
# ============================================================

returns = (
    portfolio[
        "long_short_return"
    ]
    .dropna()
)

mean_return = returns.mean()

std_return = returns.std()

total_return = (
    (1 + returns).prod()
    - 1
)

# Since observations are event-based rather than daily,
# we do not pretend these are standard daily Sharpe ratios.
# Instead, report mean return and volatility per event.
event_sharpe = (
    mean_return /
    std_return
    if std_return != 0
    else np.nan
)

win_rate = (
    (returns > 0).mean()
)

print(
    "LONG-SHORT PERFORMANCE"
)

print(
    "Number of portfolio observations:",
    len(returns)
)

print(
    f"Mean return per event: {mean_return:.4%}"
)

print(
    f"Volatility per event: {std_return:.4%}"
)

print(
    f"Total compounded return: {total_return:.4%}"
)

print(
    f"Event-level Sharpe-like ratio: {event_sharpe:.3f}"
)

print(
    f"Positive event frequency: {win_rate:.2%}"
)


# ============================================================
# CELL 38 — MAXIMUM DRAWDOWN
# ============================================================

portfolio["running_max"] = (
    portfolio["long_short_growth"]
    .cummax()
)

portfolio["drawdown"] = (
    portfolio["long_short_growth"]
    /
    portfolio["running_max"]
    - 1
)

max_drawdown = (
    portfolio["drawdown"].min()
)

print(
    f"Maximum drawdown: {max_drawdown:.2%}"
)


# ============================================================
# CELL 39 — DRAWdown PLOT
# ============================================================

plt.figure(
    figsize=(11, 5)
)

plt.plot(
    portfolio["date"],
    portfolio["drawdown"]
)

plt.axhline(
    0,
    linestyle="--"
)

plt.title(
    "Long-Short Portfolio Drawdown"
)

plt.xlabel(
    "Date"
)

plt.ylabel(
    "Drawdown"
)

plt.grid(
    alpha=0.3
)

plt.tight_layout()

plt.show()


# ============================================================
# CELL 40 — SIGNAL CORRELATIONS
# ============================================================

correlation_columns = [
    "sentiment_score",
    "sentiment_change",
    "risk_language",
    "risk_change",
    "novelty",
    "ret_5d"
]

correlations = (
    research_df[
        correlation_columns
    ]
    .corr()
)

plt.figure(
    figsize=(9, 7)
)

sns.heatmap(
    correlations,
    annot=True,
    fmt=".2f",
    center=0
)

plt.title(
    "Correlation Matrix: NLP Signals and Returns"
)

plt.tight_layout()

plt.show()


# ============================================================
# CELL 41 — RETURN DISTRIBUTION
# ============================================================

plt.figure(
    figsize=(9, 5)
)

plt.hist(
    portfolio["long_short_return"],
    bins=25
)

plt.axvline(
    0,
    linestyle="--"
)

plt.title(
    "Distribution of Long-Short Event Returns"
)

plt.xlabel(
    "Long-Short Return"
)

plt.ylabel(
    "Frequency"
)

plt.tight_layout()

plt.show()


# ============================================================
# CELL 42 — TOP AND BOTTOM SIGNAL OBSERVATIONS
# ============================================================

print("Highest NLP signal observations:")

display(
    signal_df[
        [
            "ticker",
            "date",
            "sentiment_score",
            "risk_language",
            "novelty",
            "alpha_signal",
            "ret_5d"
        ]
    ]
    .sort_values(
        "alpha_signal",
        ascending=False
    )
    .head(10)
)


print("\nLowest NLP signal observations:")

display(
    signal_df[
        [
            "ticker",
            "date",
            "sentiment_score",
            "risk_language",
            "novelty",
            "alpha_signal",
            "ret_5d"
        ]
    ]
    .sort_values(
        "alpha_signal",
        ascending=True
    )
    .head(10)
)


# ============================================================
# CELL 43 — SAVE RESEARCH DATA
# ============================================================

research_path = os.path.join(
    OUTPUT_DIR,
    "research_dataset.csv"
)

portfolio_path = os.path.join(
    OUTPUT_DIR,
    "portfolio_returns.csv"
)

signal_path = os.path.join(
    OUTPUT_DIR,
    "nlp_signals.csv"
)

research_df.to_csv(
    research_path,
    index=False
)

portfolio.to_csv(
    portfolio_path,
    index=False
)

signal_df.to_csv(
    signal_path,
    index=False
)

print("Files saved:")
print(research_path)
print(portfolio_path)
print(signal_path)


# ============================================================
# CELL 44 — SAVE REGRESSION RESULTS
# ============================================================

regression_path = os.path.join(
    OUTPUT_DIR,
    "regression_results.txt"
)

with open(
    regression_path,
    "w"
) as f:

    f.write(
        model.summary().as_text()
    )

print(
    "Regression results saved:",
    regression_path
)


# ============================================================
# CELL 45 — PROJECT SUMMARY
# ============================================================

print("=" * 70)
print("SEC FILING NLP ALPHA ENGINE — FINAL SUMMARY")
print("=" * 70)

print(
    f"Stocks analysed: {len(TICKERS)}"
)

print(
    f"Filings analysed: {len(filings)}"
)

print(
    f"Usable return observations: {len(research_df)}"
)

print(
    f"Portfolio observations: {len(portfolio)}"
)

print(
    f"Mean 5-day return: "
    f"{research_df['ret_5d'].mean():.4%}"
)

print(
    f"Mean long-short event return: "
    f"{mean_return:.4%}"
)

print(
    f"Total long-short compounded return: "
    f"{total_return:.4%}"
)

print(
    f"Maximum drawdown: "
    f"{max_drawdown:.2%}"
)

print(
    f"Positive long-short events: "
    f"{win_rate:.2%}"
)

print("=" * 70)
print("PROJECT COMPLETE")
print("=" * 70)


# ============================================================
# CELL 46 — OPTIONAL: DOWNLOAD OUTPUT FILES
# ============================================================

from google.colab import files

print(
    "Your main output files are:"
)

print(
    "1.",
    research_path
)

print(
    "2.",
    portfolio_path
)

print(
    "3.",
    signal_path
)

print(
    "4.",
    regression_path
)

# Uncomment ONE at a time if you want to download them:
#
# files.download(research_path)
# files.download(portfolio_path)
# files.download(signal_path)
# files.download(regression_path)

from google.colab import drive
drive.mount('/content/drive')
```
